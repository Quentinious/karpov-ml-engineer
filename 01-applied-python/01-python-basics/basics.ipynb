```python
def zip_(l1, l2):
    rv = []

    for i in range(min(len(l1), len(l2))):

        rv.append((l1[i], l2[i]))
    return rv
```

```python
def sum_as_ints(str_list):
    rv = 0
    for one_string in str_list:
        try:
            rv += int(one_string)
        except ValueError:
            continue
    return rv
```

```python
def find_substr(substring, string):
    win_size = len(substring)
    for i in range(len(string) - win_size + 1):
        candidate = string[i:i + win_size]
        if candidate == substring:
            return i, i + win_size
    return None

def find_substr(substring, string):
    start = string.find(substring)
    end = start + len(substring)
    return start, end
```

```python
def check_string(s):
  if s[0] == " " or s[-1] == " ":
    return False
  if not s[0].isupper():
    return False
  if  s[1:] != s[1:].lower():
    return False
  if s[-1] != ".":
    return False
  return True
```
