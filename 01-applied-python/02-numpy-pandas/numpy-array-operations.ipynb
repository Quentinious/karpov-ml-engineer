# NumPy array operations


```python
import numpy as np


def sc(vec1, vec2):
    if len(vec1) != len(vec2):
        raise ValueError(f'length are not equal: {len(vec1)} != {len(vec2)}')
    result = 0.
    for i in range(len(vec1)):
        result += vec1[i] * vec2[i]
    return result
```

```python
def normalize(X: np.ndarray) -> np.ndarray:
    min_vals = X.min(axis=0)
    max_vals = X.max(axis=0)
    range_vals = max_vals - min_vals
    return (X - min_vals) / range_vals
```

```python
X = [[1, 0], [0, 1], [1, 1]]
v = [1, 0]

import numpy as np


def cosine_similarity(X: np.ndarray, v: np.ndarray) -> np.ndarray:
    dot = X @ v
    X_norms = np.linalg.norm(X, axis=1)
    v_norm = np.linalg.norm(v)
    return dot / (X_norms * v_norm)
```

```python
x = [1, 2, 3, 4, 100, 6, 7, 8, 9]

import numpy as np


def detect_outliers(x: np.ndarray) -> np.ndarray:
    q1 = np.percentile(x, 25)
    q3 = np.percentile(x, 75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return (x < lower) | (x > upper)
```

```python
X = [[1, 2], [3, 4], [5, 6]]
lambda_ = 0.1

import numpy as np


def ridge_inverse(X: np.ndarray, lambda_: float) -> np.ndarray:
    d = X.shape[1]
    xtx = X.T @ X
    identity = np.eye(d)
    return np.linalg.inv(xtx + lambda_ * identity)
```

```python
import numpy as np


def fill_na_with_median(X: np.ndarray) -> np.ndarray:
    X_copy = X.copy()
    for i in range(X.shape[1]):
        col = X[:, i]
        median = np.nanmedian(col)
        mask = np.isnan(col)
        X_copy[mask, i] = median
    return X_copy
```

```python
categories = ["cat", "dog", "cat", "fish"]

import numpy as np


def one_hot_encode(categories: np.ndarray) -> tuple[np.ndarray, list[str]]:
    unique = sorted(set(categories))
    mapping = {cat: i for i, cat in enumerate(unique)}
    one_hot = np.zeros((len(categories), len(unique)), dtype=int)
    for idx, cat in enumerate(categories):
        one_hot[idx, mapping[cat]] = 1
    return one_hot, unique
```
