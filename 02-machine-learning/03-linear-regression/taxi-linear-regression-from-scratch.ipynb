### Собственное матричное решение

Реализуйте `LinearRegressionByMatrix(X, Y, fit_intercept=True)`. При `fit_intercept=True` добавьте столбец единиц; верните одномерный массив коэффициентов из нормального уравнения:

$$\beta=(X^T X)^{-1}X^T Y.$$

Свободный коэффициент возвращается последним, как требуется в исходной постановке.

```
import numpy as np

### Your code is here

def LinearRegressionByMatrix(X: np.ndarray, Y: np.ndarray, fit_intercept: bool = True):
    """
    :param X: матрица объектов
    :param Y: вектор ответов
    :param fit_intercept: добавить ли константную переменную в данные?

    :return: одномерный numpy-array с полученными коэффициентами beta
    """
    if fit_intercept:
        ones = np.ones((X.shape[0], 1))
        X = np.hstack((X, ones))

    xt_x = np.dot(X.T, X)
    xt_x_inv = np.linalg.inv(xt_x)
    xt_y = np.dot(X.T, Y)

    beta = np.dot(xt_x_inv, xt_y)
    return beta.flatten()
```

```
X_train = df.drop(['trip_duration', 'pickup_datetime'], axis=1).values
Y_train = df['trip_duration'].values

print(LinearRegressionByMatrix(X_train, Y_train))
```
