### Сравнение двух наборов прогнозов

В таблице поездок есть истинная длительность `trip_duration` и предоставленные прогнозы `prediction_1`, `prediction_2`. Посчитайте ошибки обеих моделей.

```
df = pd.read_csv('./taxi_dataset_with_predictions.csv', index_col=0)
```

```
df.head()
```

Сравните, насколько прогнозы двух моделей отличаются от фактической длительности.

### MSE

Вычислите среднее квадратов ошибок и сохраните результаты в `error_1` и `error_2`:

$$MSE=\frac1n\sum_{i=1}^{n}(\hat y_i-y_i)^2.$$

```
### Your code is here
error_1 = ((df['prediction_1'] - df['trip_duration']) ** 2).mean()
error_2 = ((df['prediction_2'] - df['trip_duration']) ** 2).mean()
# print(error_1,error_2)
```

```
# Выведите MSE обеих моделей (шаг 5).
print(error_1,error_2)
```

### RMSE

Возьмите квадратный корень из MSE для каждой модели:

$$RMSE=\sqrt{MSE}.$$

```
### Your code is here
rmse_error1 = error_1**(1/2)
rmse_error2 = error_2**(1/2)
```

```
# Выведите RMSE обеих моделей (шаг 6).
print(rmse_error1,rmse_error2)
```

### MAE

Рассчитайте среднюю абсолютную ошибку обеих моделей:

$$MAE=\frac1n\sum_{i=1}^{n}|\hat y_i-y_i|.$$

```
### Your code is here
error_11 = abs((df['prediction_1'] - df['trip_duration'])).mean()
error_22 = abs((df['prediction_2'] - df['trip_duration'])).mean()

# Вывод значений
# print(error_11, error_22)
```

```
# Вывод значений
print(error_11, error_22)
```

### Перепредсказания и недопредсказания

MSE и MAE одинаково оценивают ошибки +d и −d. В некоторых прикладных задачах завышение и занижение прогноза имеют разные последствия. Следующее упражнение рассматривает логарифмическую метрику и сравнение двух направлений ошибки.

### RMSLE

Сначала замените отрицательные прогнозы нулями. Затем с помощью `math` вычислите:

$$RMSLE=\sqrt{\frac1n\sum_{i=1}^{n}\left(\log(1+y_i)-\log(1+\hat y_i)\right)^2}.$$

Сохраните округлённые до трёх знаков значения в `rmsle_1`, `rmsle_2`. Рассмотрите различие штрафа за завышение и занижение прогноза.

```
import math

# 1. Заменяем все отрицательные предсказания на 0 для обеих моделей
df.loc[df['prediction_1'] < 0, 'prediction_1'] = 0
df.loc[df['prediction_2'] < 0, 'prediction_2'] = 0

# 2. Вычисляем RMSLE для первой модели (с использованием .apply для math.log)
log_actual_1 = df['trip_duration'].apply(lambda x: math.log(x + 1))
log_pred_1 = df['prediction_1'].apply(lambda x: math.log(x + 1))
rmsle_1 = round(math.sqrt(((log_actual_1 - log_pred_1) ** 2).mean()), 3)

# 3. Вычисляем RMSLE для второй модели
log_actual_2 = df['trip_duration'].apply(lambda x: math.log(x + 1))
log_pred_2 = df['prediction_2'].apply(lambda x: math.log(x + 1))
rmsle_2 = round(math.sqrt(((log_actual_2 - log_pred_2) ** 2).mean()), 3)

# Печатаем результаты для проверки
print(rmsle_1, rmsle_2)
```

```
# Место вывода RMSLE обеих моделей (шаг 9).
```

Для первой модели посчитайте число прогнозов выше и ниже фактического значения. Используйте `over_predicted_1` и `under_predicted_1`; оставьте прогнозы после замены отрицательных значений нулями.

```
# ### Your code is here
over_predicted_1 = (df['prediction_1'] > df['trip_duration']).sum()
# over_predicted_1
under_predicted_1 = (df['prediction_1'] < df['trip_duration']).sum()
print(over_predicted_1, under_predicted_1)
```
